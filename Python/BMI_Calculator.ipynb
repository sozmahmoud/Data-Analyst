```python

# BMI Calculator
Name = input("Enter Your Name Please: ")
Age = int(input("Enter Your Age Please: "))
Weight = float(input("Enter Your Weight In Kg Please: "))
Height = float(input("Enter Your Height In Cm Please: "))

Height = Height / 100
BMI = Weight / (Height ** 2)

print(f"Welcome {Name}")
print(f"Your Age Is {Age}")
print(f"Your Weight Is {Weight} Kg")
print(f"Your Height Is {Height} m")
print(f"Your BMI Is {BMI:.2f}")
```

```python
if(BMI>0):
    
    if(BMI<16):
        print(Name +" Severe Thinness")
        
    elif(BMI>=16 and BMI<17):
        print(Name +" Moderate Thinness")
    elif(BMI>=17 and BMI<18.5):
        print(Name +" Mild Thinness")
    elif(BMI>=18.5  and BMI<25):
        print(Name +" Normal")
    elif(BMI>=25 and BMI<30):
        print(Name +" Overweight")
    elif(BMI>=30 and BMI<35):
        print(Name +" Obese Class I")
    elif(BMI>=35 and BMI<40):
        print(Name +" Obese Class II")
    elif(BMI>=40):
        print(Name +" Obese Class III")

else:
    print("Check Your Detail Please:")
    
```
