```python
import os,shutil
```

```python
path=r"C:/Users/sozma/Downloads/data analyst course/python"
```

```python
file_name=os.listdir(path)
```

```python
folder_name = ["csv files", "img files", "txt files"]

for loop in range(0, 3):
    if not os.path.exists(path + "/" + folder_name[loop]):
        print(path + "/" + folder_name[loop])
        os.makedirs(path + "/" + folder_name[loop])
```

```python
for file in file_name:
    if ".csv" in file and not os.path.exists(path + "/csv files/" + file):
        shutil.move(path + "/" + file, path + "/csv files/" + file)

    elif ".jpg" in file and not os.path.exists(path + "/img files/" + file):
        shutil.move(path + "/" + file, path + "/img files/" + file)

    elif ".txt" in file and not os.path.exists(path + "/txt files/" + file):
        shutil.move(path + "/" + file, path + "/txt files/" + file)
```

```python
destination = r"C:\Users\sozma\Downloads\data analyst course\python\Building an Automated File Sorter in File Explorer"

if not os.path.exists(destination):
    os.makedirs(destination)

for folder in folder_name:
    shutil.move(path + "/" + folder, destination + "/" + folder)
```
